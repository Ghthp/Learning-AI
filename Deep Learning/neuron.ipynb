```python
import torch
import torch.nn as nn
```

```python
seed = torch.manual_seed(42)    # initial seed that helps generate the same output
inputs = torch.tensor([1.0, 2.0, 3.0])  # tensor refers to any n-dim array
```

```python
neuron = nn.Linear(in_features=3, out_features=1)
```

```python
output = neuron(inputs)
```

```python
output
```

```python
# Verifying using formula

op = neuron.weight @ inputs + neuron.bias   # @ is used to calculate the dot product
op
```
