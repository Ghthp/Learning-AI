```python
# Handling missing values

df = pd.read_csv("raw_data.csv")
df
```

```python
df.isnull()
df.isna()   # same functionality as isnull
```

```python
df.isnull().sum()   #returns the total number of missing values in each column
```

```python
# remove missing values 

df.dropna().copy()  # dropna removes all rows with missing values
df.dropna(axis = 1)  # drops columns with empty values
```

```python
# filling missing values

df.fillna(0)    # fills all missing values with 0

cleaned_data = df.copy()
age_mean = cleaned_data["age"].mean()
cleaned_data["age"] = cleaned_data["age"].fillna(age_mean)  #fills age column with mean age


cleaned_data["gender"] = cleaned_data["gender"].fillna("Unknown")

income_mean = cleaned_data["income"].mean()
cleaned_data["income"] = cleaned_data["income"].fillna(income_mean)

cleaned_data
```

```python
# forward and backward fill

cleaned_data = df.copy()
cleaned_data["income"] = cleaned_data["income"].ffill() # forward fill

cleaned_data["gender"] = cleaned_data["gender"].bfill() # backward fill

cleaned_data
```

```python
# Handling duplicate values

df.duplicated() #tells which rows are duplicated

non_duplicated_data = df.drop_duplicates().copy() # drops all duplicate rows

non_duplicated_data
```

```python
# handling datatypes
df.dtypes   # returns the datatype of columns

df2 = df.copy()
df2 = df2.fillna(0)
df2["age"] = df2["age"].astype("int64")

df2
```

```python
# Handling string datas

df["gender"].str.lower()   
df["name"].str.strip(" ")   # strips string from the separator value
df["country"].str.contains("china", case = False)   # checks if a string contains a word/string
```

```python
#Transforming our data (Feature Engineering)

# 1. apply(fnx) applies a function to a feature/ column
df2 = df.copy()

df2["tax"] = df2["income"].apply(lambda x: "10%" if x > 60000 else "20%")

# 2. map() maps the features across a column
gender_map = {"Male": "M", "Female": "F", "Unknown": "U"}
df2["gender"] = df2["gender"].map(gender_map)

# 3. assign() assigns new values from an existing column to a new column
df2 = df2.assign(new_income=df2["income"] * 1.1)

# 4. replace(old_val, new_val)
df2["country"] = df2["country"].replace("USA", "United States")

# 5. rename(columns/index = { old: new })
df2.columns = ["Id", "Name", "Age", "Country", "Gender", "Income", "Tax", "New_Income"] # changes name of all the columns
# rename() returns a new DataFrame by default, so assign it back or use inplace=True

df2 = df2.rename(columns={"Income": "Salary"})  # renames columns
# df2 = df2.rename(index={1: "First"})    # renames rows


# 6. sort_values()
# df2.sort_values("Salary", ascending=False)  # ascending=False sorts values in descending order
sorted_df = df2.sort_values(["Salary", "Age"])  # sorts based on salary but age incase incomes are same

# 7. sort_index()
sorted_df = df2.sort_index(ascending=True)

# 8. reset_index()
sorted_df.reset_index() # (drop=True) can be used if we don't want to preserve original index

# 9. Ranking
sorted_df["Salary"] = sorted_df["Salary"].fillna(25000)
sorted_df["Ranking"] = sorted_df["Salary"].rank()   # rank(method: Literal['average', 'min', 'max', 'first', 'dense']) can be used to change how we want to rank the data

sorted_df
```

```python
# Shift id column to the end
df3 = df.copy()
new_column_order = [col for col in df3 if col != "id"] + ["id"]

df3[new_column_order]

# OR we can also perform
# df3 = df.copy()
# id_col = df3.pop("id")   # removes the column together with its values
# df3["id"] = id_col      # adds it back at the end
```

```python
# Writing data in csv file

df4 = df.copy()

df4 = df4.drop_duplicates()
df4 = df4.fillna(0)
df4 = df4.sort_values("income")
df4 = df4.reset_index(drop=True)

df4.to_csv("sorted_data.csv")
```
