```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
oscar_movie_names = [
    "The Departed",
    "No Country for Old Men",
    "Slumdog Millionaire",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo",
    "12 Years a Slave",
    "Birdman",
    "Spotlight",
    "Moonlight",
    "The Shape of Water",
    "Green Book",
    "Parasite"
]

years = [2006,2007,2008,2009,2010,2011,2012,2013,2014,2015,2016,2017,2018,2019]

oscar_revenue = [291.47,171.63,378.40,49.44,484.19,133.43,232.32,187.73,103.22,98.41,65.24,195.23,321.75,258.73]

plt.plot(years,oscar_revenue) # Line plot (plt.show is used to show the graph when not using ipynb)

plt.title("Oscar winning movie revenue per year ")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
```

```python
non_oscar_movie_names = [
    "The Prestige",
    "There Will Be Blood",
    "The Dark Knight",
    "Inglourious Basterds",
    "Inception",
    "The Intouchables",
    "Django Unchained",
    "The Wolf of Wall Street",
    "Interstellar",
    "Mad Max: Fury Road",
    "La La Land",
    "Three Billboards Outside Ebbing, Missouri",
    "Avengers: Infinity War",
    "Avengers: Endgame"
]

years = [2006,2007,2008,2009,2010,2011,2012,2013,2014,2015,2016,2017,2018,2019]

non_oscar_revenue = [109.68,76.18,1008.48,321.46,839.80,426.59,426.08,407.04,764.34,380.46,523.09,162.73,2048.36,2797.50]

plt.plot(years,non_oscar_revenue) # Line plot 

plt.title("Non-Oscar winning movie revenue per year ")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
```

```python
# styling methods and attributes

plt.plot(oscar_movie_names, non_oscar_revenue) 

plt.title("Movie Revenue")
plt.xlabel("Movies")
plt.ylabel("Revenue (in $M)")

plt.style.available # can be used to view all available styles

plt.style.use("default")    # styling
plt.tight_layout()  # useful for spacing
plt.grid()  # displays a grid in the background
```

```python
# Scatter Plots
from matplotlib.lines import Line2D

names = [
    "A", "B", "C", "D", "E",
    "F", "G", "H", "I", "J",
    "K", "L", "M", "N", "O"
]

age = [
    22, 25, 29, 31, 35,
    38, 42, 45, 48, 52,
    56, 60, 64, 68, 72
]

blood_pressure = [
    118, 121, 125, 128, 130,
    126, 135, 138, 142, 145,
    150, 148, 155, 160, 165
]

colors =["green" if i < 140 else "red" for i in blood_pressure]

plt.scatter(age, blood_pressure, color=colors)

plt.title("Age VS Blood Pressure")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")

legend_items = [
    Line2D([0], [0], marker="o", color="w", label="Normal (< 140)",
           markerfacecolor="green", markersize=8),
    Line2D([0], [0], marker="o", color="w", label="Dangerous (≥ 140)",
           markerfacecolor="red", markersize=8)
]

plt.legend(handles=legend_items)

# Annotations   plt.annotate(text, xy=(x-pos, y-pos), xytext=("Text to be displayed"))
for i in range(len(names)):
    plt.annotate(names[i], xy=(age[i], blood_pressure[i]), xytext=(age[i]+0.5, blood_pressure[i]))

plt.show()
```

```python
# Pie Chart

categories = ["Food", "Rent", "Transportation", "Entertainment", "Shopping"]

expenses = [12000, 25000, 6000, 4000, 8000]

explode = [0, 0, 0, 0.2, 0]

plt.pie(expenses, labels=categories, autopct="%.1f%%",
        wedgeprops={
            "edgecolor": "black",
            "linewidth": 1,
            "linestyle": "-",
            "alpha": 1, # Transparency
            "fill": True    # colors    # and several more
        },
        explode=explode,
        startangle=90,  # rotates the chart
        shadow=True
        )  
```

```python
# Histogram - sorts numerical data into categories(bins) and how many values fall into each bin (frequency)

np.random.seed(0)
scores = np.random.normal(70, 10, 100)
range = [0, 30, 50, 70, 90] # custom ranges for the bins (0-30, 30-50, ....)

plt.hist(scores, color="green", edgecolor="black", bins=10)

plt.axvline(    # vertical line that can be used to indicate a threshold value
    x=60,
    label="Passing Marks",
    linewidth="2",
    linestyle="--",
    color="red"
)

plt.title("Student Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("No. of Students")
```

```python
# Stack plots- multiple data series are stacked on top of each other

days = ["Mon", "Tue", "Wed", "Thur", "Fri", "Sat", "Sun"]
direct = [50, 60, 70, 80, 90, 100, 105]
organic = [30, 40, 45, 50, 60, 65, 70]
social = [20, 30, 35, 50, 55, 45, 50]

plt.stackplot(days, direct, organic, social, labels=["Direct", "Organic", "Social"])

plt.title("Marketing data over the week")
plt.xlabel("Days")
plt.ylabel("No. of Customers")
plt.legend()
```

```python
# Sub Plots- allows us to draw multiple plots in the same figure
# plt.subplot(no. of rows, no. of columns, index)  index-> 1,2,3... follows row major orientation

x = [1, 2, 3, 4, 5, 6, 7]

y1 = [np.sqrt(i) for i in x] # square root
y2 = [i*2 for i in x]   # double
y3 = [i**2 for i in x]  # square
y4 = [i**3 for i in x]  # cube

plt.subplot(2, 2, 1)
plt.plot(x, y1)
plt.title("Plot1 - Square Root")

plt.subplot(2, 2, 2)
plt.plot(x, y2)
plt.title("Plot2 - Double")

plt.subplot(2, 2, 3)
plt.plot(x, y3)
plt.title("Plot3 - Square")

plt.subplot(2, 2, 4)
plt.plot(x, y4)
plt.title("Plot4 - Cube")

plt.tight_layout()
```

```python
# Object oriented approach
# We can use plt.subplots(n_rows, n_col) which returns two objects- figure (entire canvas) and axes

fig, ax = plt.subplots(2, 2)

ax[0] [0].plot(x, y1)
ax[0] [0].set_title("Square Root")

ax[0] [1].plot(x, y2)
ax[0] [1].set_title("Double")

ax[1] [0].plot(x, y3)
ax[1] [0].set_title("Square")

ax[1] [1].plot(x, y4)
ax[1] [1].set_title("Cube")

fig.suptitle("Arithmetic Operations")
fig.supylabel("Values after operations")
fig.supxlabel("Initial values")
fig.tight_layout()

# Task use loops to print the same data
```
