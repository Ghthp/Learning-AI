```python
# Seaborn is a python data visualistion tool based on matplotlib
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme() # white, whitegride, dark, darkgrid(default)
```

```python
sns.get_dataset_names()
```

```python
tips = sns.load_dataset("tips")
print(type(tips))
print(tips.head())
```

```python
# Relational plots
# 1. Scatter Plot

sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    # kind="line"   # used to set what type of plot we need (line, pie, hist)
    hue="smoker",   # highlights data points in different style (o, x)
    size="size",
    style="smoker", # changes style of data points
)
```

```python
flights = sns.load_dataset("flights")
print(flights.head())
```

```python
# 2. Line plot

sns.lineplot(       # The shaded region highlights the confidence intervals
    data= flights,
    x="year",
    y="passengers"
)
```

```python
# Categorical plots
# 1. Bar plot

sns.barplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)
```

```python
# 2. box plot

sns.boxplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)
```

```python
# Distribution plot
# Histogram

penguins = sns.load_dataset("penguins")
# print(penguins.head())

sns.histplot(
    data=penguins,
    x="body_mass_g",
    bins=30
)
```

```python
# Heatmap- shows 2d data in matrix format

flights_pivot = flights.pivot(index="month", columns="year", values="passengers")

sns.heatmap(
    data=flights_pivot,
    cmap="coolwarm",
    annot=True,     # shows the data value
    fmt="d"         # integers
)

plt.title("Flight passengers each month per year")
plt.tight_layout()
```

```python
tips = sns.load_dataset("tips")

fig, ax = plt.subplots()

sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    marker="o",
    ax=ax,
    errorbar=None   # removes confidence interval
)

ax.set_title("Total bill in a week")
ax.set_xlabel("Days")
ax.set_ylabel("Total Bill Amount")

fig.tight_layout()
```
